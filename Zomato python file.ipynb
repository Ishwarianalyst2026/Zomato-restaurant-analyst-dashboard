```python
df=pd.merge(df_data,df_country,on='Country', how= 'left')
df.head()
```

```python
# Remove duplicates
df.drop_duplicates(inplace=True)

# Fill missing cuisines
df['Cuisines'].fillna('Unknown', inplace=True)

# Fill missing city names
df['City'].fillna('Unknown', inplace=True)

# Fill missing ratings with average rating
df['Aggregate rating'].fillna(df['Aggregate rating'].mean(), inplace=True)
```

```python
print(df.isnull().sum())
print(df.duplicated().sum())
```

```python
df.drop(['Unnamed: 0'],axis=1,inplace=True)
df.columns
```

```python
df.isnull().sum()
```

```python
df.Country.value_counts()
```

```python
# Average Cost Analysis
print("Average Cost for Two:")
print(df['Average Cost for two'].mean())

# Online Delivery vs Rating
print("Average Rating by Online Delivery:")
print(df.groupby('Has Online delivery')['Aggregate rating'].mean())

# Ratings by Cuisine
print("Average Rating by Cuisine:")
print(df.groupby('Cuisines')['Aggregate rating']
      .mean()
      .sort_values(ascending=False)
      .head(10))

# Ratings by City
print("Average Rating by City:")
print(df.groupby('City')['Aggregate rating']
      .mean()
      .sort_values(ascending=False)
      .head(10))
```
